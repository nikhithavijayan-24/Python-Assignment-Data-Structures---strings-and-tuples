## Strings

String Concatenation

```python
string1 = "Hello "
string2 = input("Enter your name: ")
string = string1 + string2
print(string)
```

```python
string3= ", welcome to Python programming"
string = string1 + string2 + string3
print(string)
```

Slicing & Indexing

```python
# First Character
print(string[0])
```

```python
# Last Charcter
print(string[-1])
```

```python
# First 5 characters
print(string[ :5])
```

```python
# Last 11 characters
print(string[-11:])
```

```python
# String in reverse order
print(string[ ::-1])
```

```python
# Printing Python using slicing
string.index("Python")
print(string[23:28+1])
```

String method

```python
strM= "Python beginner tutorial"
print(str)
```

```python
# Upper case
strM.upper()
```

```python
# Lower case
strM.lower()
```

```python
# Capitalize
strM="Python beginner tutorial"
strM.capitalize()
```

```python
# Occurence of t
strM.count("t")
```

```python
# Replacing Python with Machine Learning
strM.replace("Python","Machine Learning")
```

Tuples

```python
# Tuples concatenation
tuple1 = (10,20,30)
tuple2 = (40,50,60)
t_combine = (tuple1+tuple2)
print(t_combine)
```

```python
# Repeating variable- t_combine 3 times
print(t_combine*3)
```

```python
# Accessing third element
print(t_combine[2])
```

```python
# First 3 elements
print(t_combine[:2+1])
```

```python
# Last 3 elements
print(t_combine[-3:])
```
